```python
def addition(*args):
    total = 0
    for num in args:
        total += num
    return total

print(f"addition(1, 2, 3, 40, 5): {addition(1, 2, 3, 40, 5)}")    
```

```python
def myFuction(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print(myFuction(name="John", age=30, city="New York"))
```

```python
def add(a, b):
    return a + b
type(add)  # Output: <class 'function'>
var = add
print(var(2, 3))  # Output: 5
def wrapper(func):
    return func(2, 3)
print(wrapper(add))  # Output: 5

def wrapper():
    def inner(func):
        return func(2, 3)
    return inner
print(wrapper()(add))  # Output: 5


```

```python
def succ(fuction):
    def wrapper(*args, **kwargs):
        result = fuction(*args, **kwargs)
        return result + 1
    return wrapper

@succ
def add(a, b):
    return a + b    

if __name__ == "__main__":

    #add = succ(add)
    print(add(2, 3))  # Output: 6
    print(add(2, 3))  # Output: 6
```

```python
def succ(number = 0):
    def decorator(fuction):
        def wrapper(*args, **kwargs):
            result = fuction(*args, **kwargs)
            return result + number
        return wrapper
    return decorator

#def mul_tow(x):
#    return x * 2

def mul_tow(function):
    def wrapper(*args, **kwargs):
        return function(*args, **kwargs)*2        
    return wrapper

#@succ(number=2, f=mul_tow)

#@mul_tow
@succ(number=2)
@mul_tow
def add(a, b):
    return a + b    

if __name__ == "__main__":

    #add = succ(add)
    print(add(2, 3))  # Output: 6
    print(add(2, 3))  # Output: 6
```

```python
class Count:
    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Current call count of {self.func.__name__}: {self.count}")
        return self.func(*args, **kwargs)
@Count
def add(a, b):
    return a + b
@Count
def mul(a, b):
    return a * b

if __name__ == "__main__":
    #counted_add = Count(add)
    #counted_mul = Count(mul)

    print(add(2, 3))  # Output: 5
    print(add(4, 5))  # Output: 9
    print(mul(2, 3))  # Output: 6
    print(mul(4, 5))  # Output: 20
```
