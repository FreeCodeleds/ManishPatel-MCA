```python
from collections import Counter

words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
counts = Counter(words)

print(counts.most_common(2))  # [('apple', 3), ('banana', 2)]
```

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
transposed = [num for row in zip(*matrix) for num in row]
print(f"Transposed matrix: {transposed}")  # [(1, 4, 7), (2, 5, 8), (3, 6, 9)]
```

```python
# 'name' must be positional, 'age' can be either, 'role' MUST be a keyword
def create_user(name, /, age=20, *, role="user"):
    return f"{name} ({age}) - {role}"

# Valid:
create_user("Alice", 25, role="admin")

# Error! 'name' cannot be passed as a keyword:
# create_user(name="Alice")
```

```python
# ❌ BAD: 'items' accumulates values across function calls
def add_item_bad(item, items=[]):
    items.append(item)
    return items

# ✅ GOOD: Reassign inside the function
def add_item_good(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(f"add_item_bad('apple'): {add_item_bad('apple')}  ")
print(f"add_item_good('apple'): {add_item_good('apple')}  ")
```
