```python
class A:
    def show(self): print("A")

class B(A):
    def show(self): print("B")

class C(A):
    def show(self): print("C")

class D(B, C):
    pass

print([cls.__name__ for cls in D.mro()])
# Output: ['D', 'B', 'C', 'A', 'object']
# D.show() will call B's implementation first!
```

```python
class Book:
    def __init__(self, title: str, pages: int):
        self.title = title
        self.pages = pages

    # Defines what print(book) displays
    def __str__(self):
        return f"'{self.title}' ({self.pages} pages)"

book = Book("1984", 328)
print(book)  # Output: '1984' (328 pages)
```
