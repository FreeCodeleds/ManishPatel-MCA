```python
def count_substring_occurrences(string, substring):
    """
    Count the number of occurrences of a substring in a given string.

    Args:
        string (str): The string to search within.
        substring (str): The substring to count.    
        """ 
    count = 0
    start = 0

    while True:
        start = string.find(substring, start)
        if start == -1:
            break
        count += 1
        start += len(substring)  # Move past the last found substring

    return count

if __name__ == "__main__":
    # Example usage
    main_string = "ABCDCDC"
    sub_string = "CDC"
    occurrences = count_substring_occurrences(main_string, sub_string)
    print(f"The substring '{sub_string}' occurs {occurrences} times in the main string.")
```

```python
def twoSum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target -num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i

        
```

```python
students = [{'name': 'Amit', 'age': 25}, {'name': 'Rahul', 'age': 20}]

# उम्र के आधार पर सॉर्ट करने के लिए lambda का उपयोग
students_sorted = sorted(students, key=lambda x: x['age'])

print(students_sorted)
# आउटपुट: [{'name': 'Rahul', 'age': 20}, {'name': 'Amit', 'age': 25}]
```

```python
from contextlib import contextmanager

@contextmanager
def my_context():
    print("--- काम शुरू हुआ (Setup) ---")  # with ब्लॉक में जाने से पहले चलेगा
    yield                                # यहाँ कोड एक्जीक्यूट होगा
    print("--- काम खत्म हुआ (Cleanup) ---") # with ब्लॉक से बाहर आते ही चलेगा

# इसका उपयोग:
with my_context():
    print("मैं इसके अंदर कुछ काम कर रहा हूँ...")
```

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

counter = count_up_to(5)
for number in counter:

    print(number)   
        
```

```python
for num in count_up_to(3):
    print(num)
# आउटपुट: 1, 2, 3 (एक-एक करके)
```
