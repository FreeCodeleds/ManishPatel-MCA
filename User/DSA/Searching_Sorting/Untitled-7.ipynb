```python
def count_substring_occurrences(string, substring):
    """
    Count the number of occurrences of a substring in a given string.

    Args:
        string (str): The string to search within.
        substring (str): The substring to count.    
        """ 
    count = 0
    start = 0

    while True:
        start = string.find(substring, start)
        if start == -1:
            break
        count += 1
        start += len(substring)  # Move past the last found substring

    return count

if __name__ == "__main__":
    # Example usage
    main_string = "ABCDCDC"
    sub_string = "CDC"
    occurrences = count_substring_occurrences(main_string, sub_string)
    print(f"The substring '{sub_string}' occurs {occurrences} times in the main string.")
```
